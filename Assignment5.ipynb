```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
appl = pd.read_csv('/content/drive/MyDrive/MSSP607/Data/aapl.csv')
appl
amzn = pd.read_csv('/content/drive/MyDrive/MSSP607/Data/amzn.csv')
amzn
fb = pd.read_csv('/content/drive/MyDrive/MSSP607/Data/fb.csv')
fb
goog = pd.read_csv('/content/drive/MyDrive/MSSP607/Data/goog.csv')
goog
nflx = pd.read_csv('/content/drive/MyDrive/MSSP607/Data/nflx.csv')
nflx
```

```python
appl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
nflx['ticker'] = 'NFLX'

display(appl.head())
display(amzn.head())
display(fb.head())
display(goog.head())
display(nflx.head())
```

```python
all_stocks = pd.concat([appl, amzn, fb, goog, nflx])
display(all_stocks.head())
```

```python
all_stocks.to_csv('/content/drive/MyDrive/MSSP607/Data/faanf.csv', index=False)
```

```python
all_stocks['date'] = pd.to_datetime(all_stocks['date'])
all_stocks['volume'] = all_stocks['volume'].astype(int)

display(all_stocks.info())
```

```python
all_stocks_sorted = all_stocks.sort_values(by=['date', 'ticker'])
display(all_stocks_sorted.head())
```

```python
highest_volume_rows = all_stocks_sorted.nlargest(7, 'volume')
display(highest_volume_rows)
```

```python
all_stocks_melted = pd.melt(all_stocks_sorted, id_vars=['date', 'ticker'])
display(all_stocks_melted.head())
```
