```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import numpy as np
import statistics
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/drive/MyDrive/MSSP607/Assignments/week7/InternetSales.csv',encoding='latin1')
df
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
df['Year'] = df['OrderDate'].dt.year
df['SalesAmount'] = pd.to_numeric(df['SalesAmount'], errors='coerce')
yearly_sales = df.groupby('Year')['SalesAmount'].sum().reset_index()
print(yearly_sales)
```

```python
# bar chart
plt.figure(figsize=(12, 6))
plt.bar(yearly_sales['Year'], yearly_sales['SalesAmount'],
        color='pink', edgecolor='black', width=0.6)

for i, row in yearly_sales.iterrows():
    plt.text(row['Year'], row['SalesAmount'],
             f'${row["SalesAmount"]:,.0f}',
             ha='center', va='bottom', fontsize=10)

plt.xlabel('Year', fontsize=12)
plt.ylabel('SalesAmount ($)', fontsize=12)
plt.title('Yearly Sale', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```
