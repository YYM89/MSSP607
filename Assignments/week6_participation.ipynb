```python
#Lambda
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import numpy as np
import statistics
```

```python
df = pd.read_csv('/content/drive/MyDrive/MSSP607/Assignments/week6/Student Performance Data.csv', encoding='latin1')
display(df)
```

```python
#creates a new column 'average_score'
df['average_score'] = df[['math score', 'reading score', 'writing score']].apply(lambda row: row.mean(), axis=1)
display(df.head(3))
```

```python
#creates a 'status' column. If the average_score is 60 or above, the status is 'Pass', otherwise it's 'Fail'.
passing_score = 60
df['status'] = df['average_score'].apply(lambda x: 'Pass' if x >= passing_score else 'Fail')
display(df.head(3))
```

```python
#Combine parental education fields safely
if "EDU_FATHER" in df.columns and "EDU_MOTHER" in df.columns:
    df["PARENT_EDU_COMBO"] = df.apply(
        lambda r: f"{r['EDU_FATHER']}-{r['EDU_MOTHER']}", axis=1
    )
display(df.head(5))
```
