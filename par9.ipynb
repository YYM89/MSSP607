```python
import os, requests, re, pathlib, io, time
from urllib.parse import urlencode, quote_plus
from PIL import Image
```

```python
kw = input("keyword：").strip()
n  = int(input("download amount(<=20)：") or "20")
os.makedirs("downloads", exist_ok=True)

u = "https://www.flickr.com/services/feeds/photos_public.gne?" + urlencode(
    {"format":"json","nojsoncallback":1,"tags":kw}, quote_via=quote_plus)
items = requests.get(u, timeout=30).json().get("items", [])[:n]

for i, it in enumerate(items, 1):
    url = (it.get("media") or {}).get("m");  ext = pathlib.Path(url.split("?")[0]).suffix or ".jpg"
    name = re.sub(r'[\\/:*?"<>|]+', "_", (it.get("title") or "img")).strip() or "img"
    r = requests.get(url, timeout=30);  Image.open(io.BytesIO(r.content))
    open(f"downloads/{name}_{i:03d}{ext}", "wb").write(r.content);  time.sleep(0.3)

print("Done.")
```
