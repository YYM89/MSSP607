```python
import pandas as pd
import scipy.stats as stats

# Load the dataframe
df = pd.read_csv('/content/drive/MyDrive/MSSP607/Data/Student Performance Data.csv')
print(df.head())
print(df.info())
```

```python
# Question 2 Analysis
# Group 1: Males who completed the preparation exam
group1_mask = (df['gender'] == 'male') & (df['test preparation course'] == 'completed')
group1 = df[group1_mask]

# Group 2: Others
group2 = df[~group1_mask]

# Perform t-tests for scores
scores = ['math score', 'reading score', 'writing score', 'total score']
results_q2 = {}

for score in scores:
    t_stat, p_val = stats.ttest_ind(group1[score], group2[score], equal_var=False)
    results_q2[score] = {'t_statistic': t_stat, 'p_value': p_val, 'mean_g1': group1[score].mean(), 'mean_g2': group2[score].mean()}

print("Results for Q2 (Males with Prep vs Others):")
print(pd.DataFrame(results_q2).T)

# Question 3 Analysis
# Comparison A: Gender (Male vs Female)
male = df[df['gender'] == 'male']
female = df[df['gender'] == 'female']

# Comparison B: Prep Course (Completed vs None)
prep_completed = df[df['test preparation course'] == 'completed']
prep_none = df[df['test preparation course'] == 'none']

# Comparison C: Lunch (Standard vs Free/Reduced)
lunch_std = df[df['lunch'] == 'standard']
lunch_free = df[df['lunch'] == 'free/reduced']

results_q3 = {}

# Helper function for t-test
def run_ttest(g1, g2, label_g1, label_g2, comparison_name):
    res = {}
    for score in scores:
        t_stat, p_val = stats.ttest_ind(g1[score], g2[score], equal_var=False)
        res[score] = {'t_stat': t_stat, 'p_val': p_val, f'mean_{label_g1}': g1[score].mean(), f'mean_{label_g2}': g2[score].mean()}
    return pd.DataFrame(res).T

df_gender = run_ttest(male, female, 'male', 'female', 'Gender')
df_prep = run_ttest(prep_completed, prep_none, 'completed', 'none', 'Prep Course')
df_lunch = run_ttest(lunch_std, lunch_free, 'standard', 'free/reduced', 'Lunch')

print("\nResults for Q3 - Gender:")
print(df_gender)
print("\nResults for Q3 - Prep Course:")
print(df_prep)
print("\nResults for Q3 - Lunch:")
print(df_lunch)
```

Yes, there are statistical differences, particularly in math and the total score, but the results vary by subject.

We compared Group 1 (Males who completed the test preparation course) against Group 2 (All other students).
-Math: Males who prepared scored significantly higher than the rest of the population (72.3 vs 64.8).
-Reading & Writing: There was no statistically significant difference in Reading, and the difference in Writing was marginal. This is likely because the "Others" group includes females, who generally perform better in reading and writing, balancing out the advantage that the males got from the preparation course.

Overall: The prepared males still had a significantly higher Total Score (212.3) compared to the rest of the students (201.4).
